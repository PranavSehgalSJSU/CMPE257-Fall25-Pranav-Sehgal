Write a function to perform forward propagation and back propagation to compute the gradient
on a neural network with 2 layers. Your neural network should have:
-  2 input units (d(0) = 2)
- m hidden units (d(1) = m)
- 1 output unit (d(2) = 1)

```python
"""
    File : HW1-1.ipynb
    Author(s) : Pranav Sehgal
    Description :
"""
import numpy as np

#Defines Derivative of ten
def tanhDer(x):
    return 1 - np.tanh(x)**2
```

```python
"""
    Function used for forward propogation,
    scrapped together from slides
    @param x:  input col shape
    @param W1, W2 : weight matrix of hidden and output layers
    @param b1, b2 : bias vector of hidden and output layers
    @returns: (cache, output)
"""
def forwProp(x, W1, W2, b1, b2):
    z1 = np.dot(W1, x) + b1
    a1 = np.tanh(z1)
    z2 = np.dot(W2, a1) + b2
    a2 = np.tanh(z2)

    #Get? It's funny cuz...
    cacheMap = {"x": x, "z1": z1, "a1": a1, "z2": z2, "a2": a2}
    return cacheMap, a2

"""
    Function used for backward propogation,
    scrapped together from slides
    @param y: output scalar
    @param cache: from forwProp
    returns: gradients
"""
def backProp(y, cache, W1, W2):
    x, a1, a2, z1, z2 = cache["x"], cache["a1"], cache["a2"], cache["z1"], cache["z2"]

    delta2 = (a2 - y) * tanhDer(z2)
    dW2 = np.dot(delta2, a1.T)
    db2 = delta2

    delta1 = np.dot(W2.T, delta2) * tanhDer(z1)
    dW1 = np.dot(delta1, x.T)
    db1 = delta1

    return dW1, db1, dW2, db2
```

```python
"""
    Uses Forw & back propogation w/ gradient descent to train model
    X, Y: input & tagetshape So i can remold it
    m: hidden nur cnt
    lr: laerning rate
    epochs: 
"""
def train(X, Y, m, lr, epochs):

    n = X.shape[0]
    b1 = np.zeros((m, 1))
    b2 = np.zeros((1, 1))
    W1 = np.random.randn(m, 2) * 0.3
    W2 = np.random.randn(1, m) * 0.3
    
    for epoch in range(epochs):
        loss = 0
        for i in range(n):
            x = X[i].reshape(2, 1)
            y = Y[i]

            cache, a2 = forwProp(x, W1, W2, b1, b2)
            dW1, db1, dW2, db2 = backProp(y, cache, W1, W2)

            W1 -= lr * dW1
            b1 -= lr * db1
            W2 -= lr * dW2
            b2 -= lr * db2

            loss += 0.5 * (y - a2)**2

        if epoch % 2000 == 0:
            avgLoss = loss.mean()
            print(f"Epoch {epoch}, Loss = {avgLoss}")

    return W1, b1, W2, b2
```
